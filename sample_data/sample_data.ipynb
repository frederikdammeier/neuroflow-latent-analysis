This notebook subsamples test set images/encodings/reconstructions for easier pipeline development.

```python
import numpy as np
import torch
import shutil
import os
```

```python
# generate random sample (indices in [0, 999])
sample_size = 50

rng = np.random.default_rng(seed=42)

# sample without replacement
sample_indices = np.sort(rng.choice(1000, size=sample_size, replace=False))
```

```python
# copy images
source_dir = '/u/fdammeier/data/NeuroFlow/nsd/subj01/test_img/'
target_dir = '/u/fdammeier/repositories/NeuroFlow/sample_data/images'

os.makedirs(target_dir, exist_ok=True)
for i in sample_indices:
    source_path = os.path.join(source_dir, f'{i}.png')
    target_path = os.path.join(target_dir, f'{i}.png')
    shutil.copy(source_path, target_path)
```

```python
# extract image embeddings
image_embeddings_path = '/u/fdammeier/data/NeuroFlow/nsd/subj01/nsd_sdxl_clip_test_sub1.npy'
target_image_embeddings_path = '/u/fdammeier/repositories/NeuroFlow/sample_data/sdxl_clip_embeddings.pt'

embeddings = np.load(image_embeddings_path)
sample_embeddings = embeddings[sample_indices]

# transform to torch tensor and save
sample_embeddings = torch.tensor(sample_embeddings)
torch.save(sample_embeddings, target_image_embeddings_path)
```

```python
# extract vae embeddings
vae_embeddings_path = '/u/fdammeier/generations/evals/fm-s1-d12-h13-bs24-v-cos-uni-d1664-zscore-v10-cycle-reverse-proj/sub1/single_s1_all_zfmri.pt'
target_vae_embeddings_path = '/u/fdammeier/repositories/NeuroFlow/sample_data/neurvae_embeddings.pt'

vae_embeddings = torch.load(vae_embeddings_path)
sample_vae_embeddings = vae_embeddings[sample_indices]
torch.save(sample_vae_embeddings, target_vae_embeddings_path)
```

```python
# extract fMRI zscores
fmri_zscores_path = '/u/fdammeier/data/NeuroFlow/nsd/subj01/nsd_test_fmri_zscore_sub1.npy'
target_fmri_zscores_path = '/u/fdammeier/repositories/NeuroFlow/sample_data/fmri_zscores.pt'
fmri_zscores = np.load(fmri_zscores_path)
sample_fmri_zscores = fmri_zscores[sample_indices]
# transform to torch tensor and save
sample_fmri_zscores = torch.tensor(sample_fmri_zscores)
torch.save(sample_fmri_zscores, target_fmri_zscores_path)
```

```python
# extract XFM generations
xfm_img_to_fmri_path = "/u/fdammeier/generations/evals/fm-s1-d12-h13-bs24-v-cos-uni-d1664-zscore-v10-cycle-reverse-proj/sub1/single_s1_all_sample_fmri.pt"
target_xfm_img_to_fmri_path = "/u/fdammeier/repositories/NeuroFlow/sample_data/neurovae_embeddings_from_img.pt"

xfm_img_to_fmri = torch.load(xfm_img_to_fmri_path)
sample_xfm_img_to_fmri = xfm_img_to_fmri[sample_indices]

torch.save(sample_xfm_img_to_fmri, target_xfm_img_to_fmri_path)
```

```python
xfm_fmri_to_img_path = "/u/fdammeier/generations/evals/fm-s1-d12-h13-bs24-v-cos-uni-d1664-zscore-v10-cycle-reverse-proj/sub1/single_s1_all_sample_clip.pt"
target_xfm_fmri_to_img_path = "/u/fdammeier/repositories/NeuroFlow/sample_data/img_embeddings_from_neurovae.pt"

xfm_fmri_to_img = torch.load(xfm_fmri_to_img_path)
sample_xfm_fmri_to_img = xfm_fmri_to_img[sample_indices]

torch.save(sample_xfm_fmri_to_img, target_xfm_fmri_to_img_path)
```
